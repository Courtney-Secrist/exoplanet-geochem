```python
import pandas as pd
import numpy as np
from datetime import datetime

pd.set_option('display.max_columns', 200)
print("ready")
```

```python
BASE = "https://exoplanetarchive.ipac.caltech.edu/TAP/sync"
QUERY = "select+*+from+toi"
URL = f"{BASE}?query={QUERY}&format=csv"

print("Downloading... this takes 30-60 seconds")
toi = pd.read_csv(URL)

print(f"Rows: {len(toi)}")
print(f"Columns: {len(toi.columns)}")
```

```python
today = datetime.now().strftime("%Y%m%d")
filename = f"toi_{today}.csv"
toi.to_csv(filename, index=False)
print(f"Saved {filename}")
print(f"Query used: {URL}")
```

```python
wanted = ['toi','tid','tfopwg_disp','pl_orbper','pl_rade','pl_insol','pl_eqt','st_tmag','st_dist','st_teff','st_rad']
for col in wanted:
    status = "OK" if col in toi.columns else "MISSING"
    print(f"{status:8} {col}")
```
