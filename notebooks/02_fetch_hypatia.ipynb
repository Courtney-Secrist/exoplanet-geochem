```python
rng = np.random.default_rng(42)
all_tics = toi['tid'].dropna().unique()
print("unique TIC hosts:", len(all_tics))

sample = rng.choice(all_tics, size=300, replace=False)
names = [f"TIC {int(t)}" for t in sample]
```

```python
def check(chunk):
    params = {"name": chunk, "element": ["fe"]*len(chunk)}
    r = requests.get(f"{BASE}/composition", params=params)
    if r.status_code != 200:
        print("  failed:", r.status_code)
        return []
    return r.json()

results = []
BATCH = 50
for i in range(0, len(names), BATCH):
    chunk = names[i:i+BATCH]
    results.extend(check(chunk))
    print(f"  {i+len(chunk)}/{len(names)}")

hits = [d for d in results if d.get('median_value') is not None]
rate = len(hits) / len(results)
print(f"\n{len(hits)} of {len(results)} = {100*rate:.1f}%")
print(f"projected: ~{int(rate * len(all_tics))} of {len(all_tics)} hosts")
```

```python
matched_tics = {int(d['requested_name'].replace('TIC ',''))
                for d in hits}

sub = toi[toi['tid'].isin(sample)].drop_duplicates('tid')
sub['in_hypatia'] = sub['tid'].isin(matched_tics)

print(sub.groupby('in_hypatia')['st_tmag'].describe()[['count','mean','min','max']])
```
